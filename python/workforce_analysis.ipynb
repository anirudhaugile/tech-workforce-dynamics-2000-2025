```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid") # this will make the plot look better

print("Libraries imported successfully.")
```

```python
df = pd.read_csv("../data/raw/tech_employment_2000_2025.csv")

print("Dataset loaded.")
df.head()
```

```python
df.shape # that is 532 rows and 16 columns
```

```python
df.columns
```

```python
df.info()
```

```python
yearly_summary = (
    df.groupby("year")[["new_hires", "layoffs", "net_change"]]
      .sum()
      .reset_index()
)

yearly_summary.head()
```

```python
macro_summary = (
    df.groupby("year")[["gdp_growth_us_pct", "unemployment_rate_us_pct"]]
        .mean()
        .reset_index()
)

macro_summary.head()
```

```python
yearly_summary = yearly_summary.merge(macro_summary, on="year")
yearly_summary.head()
```

```python
company_summary = (
    df.groupby("company")
      .agg({
          "new_hires": "sum",
          "layoffs": "sum",
          "hiring_rate_pct": "mean",
          "attrition_rate_pct": "mean",
          "net_change": "std",
      })
      .reset_index()
)

company_summary.head()
```

```python
company_summary.sort_values("net_change", ascending=False).head(5)
```

```python
yearly_summary.to_csv("../data/processed/yearly_summary.csv", index=False)
```
